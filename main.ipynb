```python
import pandas as pd 

df = pd.read_csv('customer_shopping_behavior.csv')

# gives 1st 5 rows of the dataframe
df.head()

# tells about dataframe like no. of rows, cols, data types
df.info()

# summary stats of all the cols not just mathematical but also categorical cols 
df.describe(include='all')

# # checking the missing values in the dataframe
# df.isnull().sum()

# Filling the missing values in the 'Review Rating' column with the median of that column grouped by 'Category'
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))

df.isnull().sum()
```

```python
# Change column names to snake case
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('(', '').str.replace(')', '')
df = df.rename(columns={'purchase_amount_usd':'purchase_amount'})
df.columns
```

```python
# Create a column age group
labels = ['Young Adult', 'Adult', 'Middle-Aged', 'Senior']
# these bins defines the age groups based on the quantiles of the age distribution by pandas qcut fxn
df['age_group'], bins = pd.qcut(df['age'], q=4, labels=labels, retbins=True)
print(df[['age','age_group']].head(10))
print('bin bndries', bins)
```

```python
# create a column for purchase_frequency_days
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
df[['frequency_of_purchases', 'purchase_frequency_days']].head(10)
```

```python
df[['discount_applied', 'promo_code_used']].head(10)
# we check if they are same or not, if same then remove either of these
# checking the entire table for similarity
df['discount_applied'].equals(df['promo_code_used'])  
df = df.drop(columns=['promo_code_used'])  
df.columns
```
